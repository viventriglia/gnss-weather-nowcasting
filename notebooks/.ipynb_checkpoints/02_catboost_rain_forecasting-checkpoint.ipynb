# CatBoost per rain forecasting: decisione e probabilità

Questo notebook affronta due obiettivi distinti sul target `is_rain` (`rain_next_hour_mm > 0.0`):

1. **modello decision-oriented**, valutato con F1 dopo aver scelto la soglia sul validation set;
2. **modello probability-oriented**, valutato con Brier score e reliability diagram.

Un blocco cronologico separato viene riservato alla futura calibrazione Venn–Abers. Il test resta sempre fuori da training, early stopping, scelta della soglia e calibrazione.

```python
from dataclasses import asdict
import json
from pathlib import Path

import catboost
from catboost import CatBoostClassifier
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'gnss_weather').exists():
    ROOT = ROOT.parent

from gnss_weather.forecasting import (
    best_f1_threshold,
    binary_metrics,
    brier_score,
    chronological_split,
    reliability_table,
)

DATASET_PATH = ROOT / 'data/derived/ml/MATE00ITA_rain_hourly_2025-08-01_2026-08-15.csv'
DATASET_REPORT = DATASET_PATH.with_suffix('.json')
MODEL_DIR = ROOT / 'data/models'
OUTPUT_DIR = ROOT / 'data/derived/ml'
MODEL_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
RANDOM_SEED = 42
catboost.__version__
```

## Caricamento e split temporale

Usiamo blocchi contigui 60% / 15% / 10% / 15%: train, validation, calibration e test. Non mescoliamo mai righe casualmente, perché rolling feature, stagionalità e persistenza atmosferica renderebbero il risultato troppo ottimistico.

```python
dataset_metadata = json.loads(DATASET_REPORT.read_text(encoding='utf-8'))
feature_columns = dataset_metadata['feature_columns']
frame = pd.read_csv(
    DATASET_PATH,
    parse_dates=['epoch_utc', 'target_window_end_utc'],
).set_index('epoch_utc').sort_index()

assert frame.index.is_monotonic_increasing
assert not frame[feature_columns + ['is_rain']].isna().any().any()
assert dataset_metadata['rain_threshold_mm_strictly_greater_than'] == 0.0

splits = chronological_split(frame)
split_summary = pd.DataFrame({
    name: {
        'start': part.index.min(),
        'end': part.index.max(),
        'rows': len(part),
        'rain_hours': int(part['is_rain'].sum()),
        'rain_fraction': part['is_rain'].mean(),
    }
    for name, part in splits.items()
}).T
split_summary
```

```python
def xy(name):
    part = splits[name]
    return part[feature_columns], part['is_rain'].astype(int)

X_train, y_train = xy('train')
X_validation, y_validation = xy('validation')
X_calibration, y_calibration = xy('calibration')
X_test, y_test = xy('test')

common_parameters = dict(
    iterations=1000,
    depth=6,
    learning_rate=0.03,
    l2_leaf_reg=5.0,
    loss_function='Logloss',
    random_seed=RANDOM_SEED,
    allow_writing_files=False,
    thread_count=-1,
)
```

## 1. Modello binario ottimizzato per F1

Qui usiamo `auto_class_weights='Balanced'` per dare più peso alle ore piovose. Le sue uscite sono trattate come **score decisionali**, non come probabilità calibrate. La soglia non viene scelta sul test, ma sul validation set.

```python
decision_model = CatBoostClassifier(
    **common_parameters,
    auto_class_weights='Balanced',
    eval_metric='AUC',
)
decision_model.fit(
    X_train,
    y_train,
    eval_set=(X_validation, y_validation),
    early_stopping_rounds=100,
    verbose=100,
)
decision_model.save_model(MODEL_DIR / 'rain_catboost_decision.cbm')
```

```python
validation_decision_score = decision_model.predict_proba(X_validation)[:, 1]
selected_threshold = best_f1_threshold(y_validation, validation_decision_score)
test_decision_score = decision_model.predict_proba(X_test)[:, 1]
test_binary = binary_metrics(y_test, test_decision_score, selected_threshold.threshold)

pd.DataFrame(
    [asdict(selected_threshold), asdict(test_binary)],
    index=['validation_threshold_selection', 'test_evaluation'],
)
```

```python
threshold_grid = np.linspace(0.01, 0.99, 199)
f1_grid = [binary_metrics(y_validation, validation_decision_score, t).f1 for t in threshold_grid]
fig, axis = plt.subplots(figsize=(9, 4), constrained_layout=True)
axis.plot(threshold_grid, f1_grid, color='tab:blue')
axis.axvline(selected_threshold.threshold, color='tab:red', linestyle='--', label=f'soglia={selected_threshold.threshold:.3f}')
axis.set(xlabel='Soglia decisionale', ylabel='F1 validation', title='Scelta della soglia senza usare il test')
axis.grid(alpha=0.25)
axis.legend()
plt.show()
```

## 2. Modello probabilistico valutato con Brier score

Per le probabilità eliminiamo i class weights. Un primo fit usa il validation set soltanto per scegliere il numero di alberi; poi il modello finale viene riaddestrato su train+validation con quel numero fissato. Il calibration set resta escluso ed è pronto per Venn–Abers.

```python
probability_tuning_model = CatBoostClassifier(
    **common_parameters,
    eval_metric='BrierScore',
)
probability_tuning_model.fit(
    X_train,
    y_train,
    eval_set=(X_validation, y_validation),
    early_stopping_rounds=100,
    verbose=100,
)
best_iteration = max(1, probability_tuning_model.get_best_iteration() + 1)
best_iteration
```

```python
X_proper_train = pd.concat([X_train, X_validation])
y_proper_train = pd.concat([y_train, y_validation])
final_probability_parameters = {
    **common_parameters,
    'iterations': best_iteration,
    'eval_metric': 'BrierScore',
}
probability_model = CatBoostClassifier(**final_probability_parameters)
probability_model.fit(X_proper_train, y_proper_train, verbose=False)
probability_model.save_model(MODEL_DIR / 'rain_catboost_probability_raw.cbm')

calibration_probability = probability_model.predict_proba(X_calibration)[:, 1]
test_probability = probability_model.predict_proba(X_test)[:, 1]
test_brier = brier_score(y_test, test_probability)
climatology_probability = float(y_proper_train.mean())
climatology_brier = brier_score(y_test, np.full(len(y_test), climatology_probability))
brier_skill = 1.0 - test_brier / climatology_brier

pd.Series({
    'best_iterations': best_iteration,
    'proper_train_climatology': climatology_probability,
    'test_brier_raw_catboost': test_brier,
    'test_brier_climatology': climatology_brier,
    'brier_skill_vs_climatology': brier_skill,
})
```

```python
reliability = reliability_table(y_test, test_probability, bins=10)
fig, axis = plt.subplots(figsize=(6, 6), constrained_layout=True)
axis.plot([0, 1], [0, 1], color='black', linestyle=':', label='calibrazione ideale')
axis.plot(
    reliability['predicted_probability'],
    reliability['observed_frequency'],
    marker='o',
    color='tab:blue',
    label='CatBoost raw',
)
for row in reliability.itertuples():
    axis.annotate(str(row.count), (row.predicted_probability, row.observed_frequency), xytext=(4, 4), textcoords='offset points', fontsize=8)
axis.set(xlim=(0, 1), ylim=(0, 1), xlabel='Probabilità prevista', ylabel='Frequenza osservata', title='Reliability diagram sul test')
axis.grid(alpha=0.25)
axis.legend()
plt.show()
reliability
```

Il reliability diagram è diagnostico: le probabilità qui sono ancora grezze. Nel prossimo notebook Venn–Abers verrà adattato agli score del solo blocco `calibration` e valutato una volta sul test.

```python
importance = pd.Series(
    probability_model.get_feature_importance(),
    index=feature_columns,
).sort_values(ascending=False)
fig, axis = plt.subplots(figsize=(9, 6), constrained_layout=True)
importance.head(15).sort_values().plot.barh(ax=axis, color='tab:green')
axis.set(xlabel='CatBoost feature importance', title='Prime 15 feature del modello probabilistico')
axis.grid(axis='x', alpha=0.25)
plt.show()
importance.head(15).to_frame('importance')
```

## Salvataggio degli artefatti

Salviamo separatamente score di calibration e test. In questo modo il prossimo passo potrà calibrare su `calibration` e confrontare probabilità raw/Venn–Abers sul medesimo test senza riaddestrare di nascosto.

```python
def prediction_frame(split_name, part, decision_score, probability):
    return pd.DataFrame({
        'epoch_utc': part.index,
        'split': split_name,
        'is_rain': part['is_rain'].to_numpy(dtype=int),
        'rain_next_hour_mm': part['rain_next_hour_mm'].to_numpy(),
        'decision_score': decision_score,
        'decision_prediction': (decision_score >= selected_threshold.threshold).astype(int),
        'probability_raw': probability,
    })

calibration_decision_score = decision_model.predict_proba(X_calibration)[:, 1]
predictions = pd.concat([
    prediction_frame('calibration', splits['calibration'], calibration_decision_score, calibration_probability),
    prediction_frame('test', splits['test'], test_decision_score, test_probability),
], ignore_index=True)
predictions_path = OUTPUT_DIR / 'MATE00ITA_catboost_calibration_test_predictions.csv'
predictions.to_csv(predictions_path, index=False)

metrics = {
    'dataset': str(DATASET_PATH.relative_to(ROOT)),
    'catboost_version': catboost.__version__,
    'random_seed': RANDOM_SEED,
    'feature_columns': feature_columns,
    'split': {
        name: {
            'start': part.index.min().isoformat(),
            'end': part.index.max().isoformat(),
            'rows': len(part),
            'rain_hours': int(part['is_rain'].sum()),
        }
        for name, part in splits.items()
    },
    'decision_model': {
        'validation_selection': asdict(selected_threshold),
        'test': asdict(test_binary),
        'model_path': 'data/models/rain_catboost_decision.cbm',
    },
    'probability_model_raw': {
        'best_iterations': best_iteration,
        'test_brier': test_brier,
        'test_climatology_brier': climatology_brier,
        'brier_skill_vs_climatology': brier_skill,
        'model_path': 'data/models/rain_catboost_probability_raw.cbm',
    },
    'calibration_policy': 'calibration block reserved for future Venn-Abers; test untouched',
}
metrics_path = OUTPUT_DIR / 'MATE00ITA_catboost_metrics.json'
metrics_path.write_text(json.dumps(metrics, indent=2) + '\n', encoding='utf-8')
predictions_path, metrics_path
```

## Limiti da mantenere espliciti

- Un solo anno e una sola stazione non dimostrano generalizzazione climatica.
- Il test è soprattutto estivo e ha una frequenza di pioggia inferiore al train: è un vero distribution shift stagionale.
- Il target è pioggia ERA5 sulla cella, non un pluviometro in giardino.
- F1 dipende dalla soglia; Brier valuta le probabilità senza scegliere una soglia.
- Venn–Abers richiederà attenzione all'ipotesi di exchangeability per una serie temporale.
